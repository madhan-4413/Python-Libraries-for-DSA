### Aim:

To study and practice various NumPy operations, including array creation, attributes, indexing, slicing, broadcasting,

<b>Case Study</b>: A supermarket store manager wants to analyze product sales over 4 weeks in January from their three branches (branch A,
B, and c). Using NumPy, extract meaningful insights from the raw sales data.



**Step 1:** Install and import the NumPy Library using the following commands:

pip install numpy


```python
#importing numpy
import numpy as np
```

**Step 2:** Create arrays for three branches, a, b, and c, for four weeks of store sales

```python
#weeks[1,2,3,4]
branch_a = np.array([120,150,190,125])
branch_b = np.array([130,160,280,175])
branch_c = np.array([140,150,250,180])
```

```python
#created arrays
branch_a, branch_b, branch_c
```

**Step 3:** Check the properties of the arrays with different attributes


```python
#dimensions of the array
branch_c.ndim
```

```python
#shape of the array 
branch_b.shape
```

```python
#size of the array
branch_a.size
```

```python
#Datatype of the branches
branch_c.dtype
```

```python
#space used by each element in the array
branch_a.itemsize, branch_c.itemsize
```

**Step 4:** Retretive Sales data by Indexing & Slicing the arrays


```python
#Index starts with 0
#Get the branch_b week 2 sales data
branch_b[1]
```

```python
#Get the unique values in the branch_b 
np.unique(branch_c)
```

```python
#Get branch_b week 1,2 sales data
branch_a[0:3]
```

```python
#Get branch_c last two weeks sales data
branch_c[2:5]
```

```python
#Get branch_a last three weeks sales data (negitive slicing)
branch_a[-3:]
```

**Step 5:** Assign new values to the branch sales (Broadcasting)


```python
branch_b
```

```python
#Make a new copy for branch_b
branch_b_sunil= branch_b
```

```python
branch_b_sunil
```

```python
#updating week 1sales data 
branch_b_sunil[0] = 125
branch_b_sunil
```

```python
#Adding week 5,6 sales data to the branch_b_sunil array
branch_b_sunil = np.append(branch_b_sunil,[230,185])
```

```python
branch_b_sunil
```

```python
# Updating week 5, 6 sales data using slicing
branch_b_sunil[4:6] = [300, 170]
branch_b_sunil
```

**Step 6:** Arithmetic Operations on the sales data

```python
branch_c
```

```python
#Adding +2sales to all 4 weeks data
branch_a + 2
```

```python
#Subtracting 100 sales from the week 4 data with indexing in branch_c
branch_c[3] - 10
```

```python
#Double the sales in week 2 with indexing in branch_c
branch_c[1] * 2
```

```python
#Half sales in week 2 & 3 
branch_c[1:3]  / 2
```

**Step 7:** Statistical Operations on the sales data


```python
branch_a, branch_b, branch_c
```

```python
#What are the Total sales of brranch_b
np.sum(branch_b)
```

```python
#What are the Max and min sales of the branch_b
np.max(branch_b), np.min(branch_b)
```

```python
#What are the average (mean) sales of the each branch 
print(np.mean(branch_b), np.mean(branch_c))
```

**Step 8:** Combine all branches sales data (Concatenating)

```python
#Combaing all 3 branchs sales 
all_branches_sales = np.concatenate([branch_a, branch_b, branch_c])
```

```python
branch_a.ndim
```

```python
all_branches_sales
```

```python
#Total sales of all branchs
np.sum(all_branches_sales)
```

**Step 9:** Reshape the all_branches_sales into 3x4 matrix


```python
all_branches_sales.ndim, all_branches_sales.shape
```

```python
all_branches_sales = all_branches_sales.reshape(3,4)
all_branches_sales
```

```python
#Indexing 2-D array
all_branches_sales[0]
```

```python
#sort Row wise sales
np.sort(all_branches_sales, axis=1)
```

```python
# Splitting the all_branches_sales array into 3 different arrays
# Splitting based on rows (axis=0)
split_a, split_b, split_c = np.split(all_branches_sales, [1,2], axis=0)
```

```python
split_a
```

```python
split_b
```

```python
# What are the Week 2 sales of all three branches?
# Slice only column 2
week_2_sales = all_branches_sales[:3, 1:2]
```

```python
print(week_2_sales)
```
