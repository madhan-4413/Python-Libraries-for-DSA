```python
# List of columns with score
Total_score = ['first_score','second_score']
```

```python
#Rounding Attendance to 2 Decimal Points
df_sunil['avg_run_rate_match'] = df_sunil['avg_run_rate_match'].round(2)
df_sunil.head(2)
```

**Step 5:** Conditional Filtering

```python
# Top 2 Teams with avg_run_rate_match >8
df_sunil[df_sunil['avg_run_rate_match'] > 8].head(2)
```

```python
# Teams  from 'Delhi' OR from 'chennai'
df_sunil[(df_sunil['team1'] == 'Delhi') | (df_sunil['team2'] == 'Chennai')].head()
```

```python
#Teams who winners and they must be from runrate >8
df_sunil[(df_sunil['avg_run_rate_match'] >=8 ) & (df_sunil['winner'] == 'Chennai' )]
```

```python
# Filtering with Multiple Cities using .isin()
df_sunil[df_sunil['winner'].isin(['Delhi', 'Hyderabad', 'Bangalore'])].head()
```

**Step 6:** Sorting Table

```python
#Sorting Table based on a single column - avg_run_rate_match  & Match_Totalscore
df_sunil.sort_values(['avg_run_rate_match', ], ascending=False).head()
```

**Step 7:** Updating Data in the Table

```python
#Updating Hyderabad as Hyd in winner Column
df_sunil['winner'] = df_sunil['winner'].replace(['Hyderabad'],['Hyd'])
```

```python
df_sunil[df_sunil['winner']=='Hyd'].head()
```

```python
# Updating Match total score 23 to 550
df_sunil.loc[23, 'Match_Totalscore'] = 550
print(df_sunil.loc[23]['Match_Totalscore'])
```

```python
#Changing the Column Name
df_sunil = df_sunil.rename(columns={'first_score': 'firstinnings_score', 'second_score':'secondinnings_score', })
df_sunil.head(1)
```
