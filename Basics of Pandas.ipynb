```python
#importing Numpy & pandas
import numpy as np
import pandas as pd
```

**Step 2:** Create a 2-D array


```python
# Considering a 4-week bike sales data in three different branches
# rows --> branches, columns --> weeks
sales = np.array([[160, 240, 176, 230],
                  [170, 120, 145, 280],
                  [168, 235, 175, 275]])
```

```python
#sales array properties
sales.shape, sales.ndim
```

**Step 3:** Converting the created 2-D array into DataFrame using Pandas

```python
df = pd.DataFrame(sales)
df
```

```python
#creating the index & columns to the Data Frames
#Indexing--> Branchs, Columns-->weeks
df_index = ['Branch_1', 'Branch_2', 'Branch_3']
df_columns = ['week_1', 'week_2','week_3', 'week_4']
```

```python
#Assing the index & columns to the dataframe
df = pd.DataFrame(data=sales, index=df_index, columns=df_columns)
```

```python
#created dataframes
df
```

```python
# Adding Week_5 Data (Column)
# Must match number of rows (3 branches)
df['week_5'] = [210, 220, 230]
df
```

```python
# Add Branch_4 (Row)

branch_4 = pd.DataFrame({
 'week_1': [155],
 'week_2': [205],
 'week_3': [175],
 'week_4': [210],
 'week_5': [225]
}, index=['Branch_4'])

df = pd.concat([df, branch_4])
df
```
